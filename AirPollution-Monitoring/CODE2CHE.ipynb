```
from datetime import datetime
import pandas as pd
import datawig  #for imputation
import numpy as np
```

```
#Reading code-1 file
chennai_new_df = pd.read_csv('/content/chennai_final_raw_sat_data_modis.csv')
chennai_new_df.head()
```

```
#creating a time stamp for easy viewing of date and time
chennai_new_df["time_stamp"] = ''
i=0
for i in range(len(chennai_new_df)):
    chennai_new_df.iloc[i,13] = datetime(int(chennai_new_df.iloc[i,1]),int(chennai_new_df.iloc[i,2]),
                                       int(chennai_new_df.iloc[i,3]), int(chennai_new_df.iloc[i,4]),
                                       int(chennai_new_df.iloc[i,5]),int(chennai_new_df.iloc[i,6]))
```

```
# Marking respective regions based on lat long extent
def places_satellite_data(str, lat1,lat2,lon1,lon2):
    region = chennai_new_df[(chennai_new_df['Latitude'] > lat1) & (chennai_new_df['Latitude'] < lat2) & (chennai_new_df['Longitude'] > lon1) & (chennai_new_df['Longitude'] < lon2)]
    region['Place'] = str
    return region
```

```
#calling 'places_satellite_data'  function
Alandur     = places_satellite_data('Alandur', 12.99,13.01,80.18,80.21)
Velachery   = places_satellite_data('Velachery', 12.96,13.00, 80.20,80.23)
Manali      = places_satellite_data('Manali', 13.14, 13.21, 80.24, 80.30)
```

```
#merging all dataframes together
dflist = [Alandur, Velachery, Manali ]
chennai_sat_data_df = pd.DataFrame()
for f in dflist:
    chennai_sat_data_df = chennai_sat_data_df.append(f)
chennai_sat_data_df.head()
```

### Miscellaneous operations

```
#for similar time intervals as ground data i.e. for every 15 mins
chennai_sat_data_df['RoundedMinute'] = ""
#to round satellite time in minutes to every 15mins similar to cpcb
chennai_sat_data_df['RoundedMinute'] = pd.cut(x=chennai_sat_data_df['Minute'], bins=[0, 15, 30, 45, 59], labels=['0','15','30','45'])
#for minutes with 0 fill rounded minutes with 0
chennai_sat_data_df['RoundedMinute'].replace(np.nan, 0, inplace=True)
chennai_sat_data_df['RoundedMinute'] = chennai_sat_data_df['RoundedMinute'].astype(int)
```

```
#sorting values according to date   
chennai_sat_data_df = chennai_sat_data_df.sort_values(['Year', 'Month', 'Day', 'Hour', 'RoundedMinute','Place'], ascending=[True, True, True, True, True, True])
#rearranging columns
chennai_sat_data_df = chennai_sat_data_df.reindex(columns=['Year','Month','Hour','Day','RoundedMinute','Latitude','Longitude', 'Place',
                                                           'time_stamp','updated_AOD_Land_and_Ocean_datawig','Optical_Depth_Land_And_Ocean',
                                                           'Image_Optical_Depth_Land_And_Ocean','Land_sea_Flag','Land_Ocean_Quality_Flag'])
```

```
#reindexing dataframe
chennai_sat_data_df.reset_index(drop=True, inplace=True)
```

```
#to get number of pixels for each day(same hour and minute) for a particular place 
pixel_count_sat_data = [] 
pixel_count_sat_data = chennai_sat_data_df.groupby(['Year','Month', 'Day','Hour','RoundedMinute','Place']).size().reset_index(name='count')
```

### Imputation of Image Optical Depth

```
#1. Dividing into year-wise data for imputation
def year_wise(str, year):
     return chennai_sat_data_df[chennai_sat_data_df[str] == year]  
yr_2016 = year_wise('Year', 2016)
yr_2017 = year_wise('Year', 2017)
yr_2018 = year_wise('Year', 2018)
yr_2019 = year_wise('Year', 2019)
yr_2020 = year_wise('Year', 2020)
```

```
# 2.IMPUTATION
#df - yearwise dataframe to be imputed, str - column name
#month_num - month number, null_value - null value that is to be refilled 
#Imputation is done for every month, in a certain place, based on weights(distance) of latitude and longitude 
#Haversine formula is not included to make things simpler
```

```
def imputation_AOD(df):
    
    def month_wise_df(df, month_num, str, null_value):
        df_mon = df[df['Month'] == month_num]
        
        x_columns = ['Latitude', 'Longitude', 'Place', 'Year']
        
        value = df_mon[df_mon[str] != null_value]
        no_value = df_mon[df_mon[str] == null_value]
        
        imputer = datawig.SimpleImputer(
            input_columns = x_columns, # column(s) containing information about the column we want to impute
            output_column = str, # the column we'd like to impute values for
            output_path = 'datawig_imputation' # stores model data and metrics
            )
        imputer.fit(train_df = value, num_epochs = 120)
        imputed = imputer.predict(no_value)
        imputed.head()
        
        datawig_imputation_data = list()
        predictions = np.asarray(imputed['Image_Optical_Depth_Land_And_Ocean_imputed'])
        cnt = 0
        for i in range(df_mon.shape[0]):
            if(df_mon.iloc[i][str] == null_value):
                datawig_imputation_data.append(predictions[cnt])
                cnt+=1
            else:
                datawig_imputation_data.append(df_mon.iloc[i][str])
        print(cnt,"rows have been imputed")
        df_mon['updated_AOD_Land_and_Ocean_datawig'] = np.asarray(datawig_imputation_data)
        
        return df_mon
        
    mon_1 = month_wise_df(df, 1, 'Image_Optical_Depth_Land_And_Ocean', -9999)
    mon_2 = month_wise_df(df, 2, 'Image_Optical_Depth_Land_And_Ocean', -9999)
    mon_3 = month_wise_df(df, 3, 'Image_Optical_Depth_Land_And_Ocean', -9999)
    
    month_dflist = [mon_1, mon_2, mon_3]
    imputed_monthwise_df = pd.DataFrame()
    for mon in month_dflist:
        imputed_monthwise_df = imputed_monthwise_df.append(mon)
        
    return imputed_monthwise_df
```

```
#calling Imputation AOD function
yr_2016_imputed = imputation_AOD(yr_2016)
yr_2017_imputed = imputation_AOD(yr_2017)
yr_2018_imputed = imputation_AOD(yr_2018)
yr_2019_imputed = imputation_AOD(yr_2019)
yr_2020_imputed = imputation_AOD(yr_2020)
```

```
# 3.Appending them back
imputed_year_dflist = [yr_2016_imputed, yr_2017_imputed, yr_2018_imputed,
                       yr_2019_imputed, yr_2020_imputed ]
chennai_sat_data_imputed = pd.DataFrame()
for imputed_df in imputed_year_dflist:
    chennai_sat_data_imputed = chennai_sat_data_imputed.append(imputed_df)
```

cross checking imputed values and observed values by their mean values
